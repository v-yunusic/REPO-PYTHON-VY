```python
df['nombre_prod'] = df['nombre_prod'].str.upper()
df['nombre_prod'] = df['nombre_prod'].str.strip()
df['nombre_prod'] = df['nombre_prod'].fillna('DESCONOCIDO')
#limpieza caracter basura productos
df['nombre_prod'] = df['nombre_prod'].str.replace(r'[^a-zA-Z0-9 ]', '', regex=True)#Formateo de los caracteres basura
df['nombre_prod'] = df['nombre_prod'].str.replace(r'\s+', ' ', regex=True)

set(df['nombre_prod'].unique().tolist())
```

```python
productos = set(df['nombre_prod'].unique().tolist())
matches = set()

for name in productos:
  max_rtio = 0
  for mtch in productos:
    if name != mtch:
      rtio = fuzz.ratio(name, mtch)
      if rtio > 93 and rtio > max_rtio:
        if name not in matches or mtch not in matches:
          print(f'{name} -> {mtch} ({rtio:.2f}%)')
          max_rtio = rtio
          matches.add(name)
          matches.add(mtch)
          df.loc[df['nombre_prod'] == name, 'nombre_prod'] = mtch
```
